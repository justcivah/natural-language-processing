```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch
import os
```

```python
def run_inference(model_name, training_domain, input, max_length=192):
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    model_path = os.path.abspath(f"../models/google/{model_name}/student_{training_domain}/best-model")

    tokenizer = AutoTokenizer.from_pretrained(model_path)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_path).to(device)

    encoded = tokenizer.encode(input, return_tensors='pt', padding=True, truncation=True, max_length=max_length)
    generated = model.generate(encoded, max_length=max_length)
    decoded = tokenizer.decode(generated, skip_special_tokens=True)

    return decoded
```

```python
# base model
# options are: 't5-efficient-small', 't5-efficient-base', 'mt5-small', 'mt5-base' 
model_name = 'mt5-base'

# domain used to train the model
# options are: 'a', 'b', 'ab'
training_domain = 'ab'

# input of the model
input = 'I sometimes wonder whether i should quit university'
# input = 'This work investigates about black-box knowledge distillation from a GPT-5 Mini teacher for language translation tasks, by training four student models from the T5-Efficient and mT5 architectures on two different domains.'
# input = "Sophia feels cute today" # carina
# input = "Tom feels cute today" # carino

output = run_inference(model_name, training_domain, input)
print(output)
```
